## T Test

### One Sample T Test

$$
t = \frac{\bar{x} - \mu}{s_{\bar{x}}}
$$

где

$$
s_{\bar{x}} = \frac{s}{\sqrt{n}}
$$

$\mu$ - Proposed constant for the population mean
$\bar{x}$ - Sample mean
$n$ - Sample Size
$s$ - Sample standard deviation
$$\frac{s}{\sqrt{n}$$ - Estimated standard error of the mean (s/sqrt(n))

```python
ages = [10,20,35,50,28,40,55,18,16,55,30,25,43,18,30,28,14,24,16,17,32,35,26,27,65,18,43,23,21,20,19,70]
```

```python
len(ages)
```

```python
import numpy as np
ages_mean = np.mean(ages)
print(ages_mean)
```

```python
# Lets take sample

sample_size = 10
age_sample = np.random.choice(ages, sample_size)
```

```python
age_sample
```

```python
from scipy.stats import ttest_1samp
```

```python
ttest, p_value = ttest_1samp(age_sample, 30)
```

```python
print(p_value)
```

```python
if p_value < 0.05: # alpha value is 0.05 or 5%
    print(' We are rejecting null hypothesis')
else:
    print('We are accepting null hypothesis')
```

## Some more Examples

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import math
np.random.seed(6)
school_ages = stats.poisson.rvs(loc = 18, mu = 35, size = 1500)
classA_ages = stats.poisson.rvs(loc = 18, mu = 30, size = 60)
```

```python
classA_ages.mean()
```

```python
_, p_value = stats.ttest_1samp(a = classA_ages, popmean = school_ages.mean())
```

```python
school_ages.mean()
```

```python
if p_value < 0.05: # alpha value is 0.05 or 5%
    print(' We are rejecting null hypothesis')
else:
    print('We are accepting null hypothesis')
```

## Two-sample T-test With Python

$$
t = \frac{\bar{x}_1 - \bar{x}_2}
{\sqrt{s^2\left(\frac{1}{n_1} + \frac{1}{n_2}\right)}}
$$

$$
s^2 =
\frac{
\displaystyle\sum_{i=1}^{n_1}(x_i-\bar{x}_1)^2
+
\displaystyle\sum_{j=1}^{n_2}(x_j-\bar{x}_2)^2
}
{n_1+n_2-2}
$$

```python
np.random.seed(12)
ClassB_ages = stats.poisson.rvs(loc = 18, mu = 33, size = 60)
ClassB_ages.mean()
```

```python
_, p_value = stats.ttest_ind(a = classA_ages, b = ClassB_ages, equal_var = False)
```

```python
if p_value < 0.05: # alpha value is 0.05 or 5%
    print(' We are rejecting null hypothesis')
else:
    print('We are accepting null hypothesis')
```

## Paired T-test With Python

```python
weight1 = [25, 30, 28, 35, 38, 34, 26, 29, 30, 26, 28, 32, 31, 30, 45]
weight2 = weight1 + stats.norm.rvs(scale = 5, loc = 1.25, size = 15)
```

```python
print(weight1)
print(weight2)
```

```python
weight_df = pd.DataFrame({'weight_10': np.array(weight1),
                          'weight_20': np.array(weight2),
                         "weight_change":np.array(weight2) - np.array(weight1)})
```
